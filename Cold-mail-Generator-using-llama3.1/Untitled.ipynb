```python
# Chromadb is used beacuse - it does a semantic search in the document and it works on euclidian distance
import chromadb
client = chromadb.Client()
collection = client.create_collection(name="my_collection")
```

```python
collection.add(
    documents=[
        " My name is Indraneel"

    ],
    ids = ['id1']
)   
```

```python
all_docs = collection.get()
all_docs
```

```python
collection.delete(ids=all_docs['ids'])
collection.get()
```
